```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
url = "https://vincentarelbundock.github.io/Rdatasets/csv/AER/TeachingRatings.csv"
```

```python
df = pd.read_csv(url)
```

```python
print("Dataset shape:", df.shape)
```

```python
print(df.head())
```

(Q1)  Duplicate cases using prof

```python
# Find duplicate cases
duplicates = df[df.duplicated(subset="prof", keep=False)]
print("\nDuplicate cases:")
print(duplicates)
```

This helped me find and print all the rows in my dataframe where the values in the "prof" column are repeated.

```python

print("\nTotal observations:", len(df))
```

```python

print("Unique instructors:", df["prof"].nunique())
```

```python

mean_age = df["age"].mean()
```

```python

std_age = df["age"].std()

print("\nUsing all observations:")
print("Average age =", mean_age)
print("Standard deviation =", std_age)
```

```python

df_unique = df.drop_duplicates(subset="prof", keep="first")

print("\nAfter removing duplicate instructors:")
print("Number of observations:", len(df_unique))

mean_age_unique = df_unique["age"].mean()
std_age_unique = df_unique["age"].std()

print("Average age =", mean_age_unique)
print("Standard deviation =", std_age_unique)
```

(Q2) Bar chart for division and evaluation

```python

avg_eval_division = df.groupby("division")["eval"].mean()

print("\nAverage evaluation by division:")
print(avg_eval_division)

plt.figure(figsize=(7,5))

plt.bar(
    avg_eval_division.index,
    avg_eval_division.values,
    color=["skyblue", "lightcoral"],
    edgecolor="black"
)

plt.xlabel("Course Division")
plt.ylabel("Average Teaching Evaluation")
plt.title("Average Teaching Evaluation by Course Division")

plt.show()
```

I'm grouping my dataframe by the course division and calculating the average evaluation score for each one.

Then printing out the results.

After that I'm plotting a bar chart to visualize these averages setting up the custom colors, labels, and title to make the graph clear.

(Q3) Age vs Evaluation

```python
import numpy as np
import matplotlib.pyplot as plt

correlation = df["age"].corr(df["eval"])

print("Correlation between age and evaluation =", correlation)

plt.figure(figsize=(8,6))

# Scatter plot
plt.scatter(
    df["age"],
    df["eval"],
    alpha=0.6,
    color="skyblue",
    edgecolor="black"
)

# Calculate regression line
slope, intercept = np.polyfit(df["age"], df["eval"], 1)

# Create regression line
x = np.array([df["age"].min(), df["age"].max()])
y = slope * x + intercept

plt.plot(
    x,
    y,
    color="red",
    linewidth=2,
    label="Regression Line"
)

plt.xlabel("Instructor Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Relationship Between Age and Teaching Evaluation")
plt.legend()

plt.show()
```

I'm calculating the correlation coefficient between the instructor's age and their evaluation score and printing out the result.

Then, I'm creating a scatter plot to visualize the data and adding a fitted red regression line to show the overall trend complete with custom labels a title and a legend.

(Q4) Gender differentiated scatterplots

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

# ---------------- FEMALE ----------------
plt.subplot(1,2,1)

female = df[df["gender"] == "female"]

plt.scatter(
    female["age"],
    female["eval"],
    alpha=0.6,
    color="hotpink",
    edgecolor="black"
)

# Regression line for female
slope_f, intercept_f = np.polyfit(
    female["age"],
    female["eval"],
    1
)

x_f = np.array([
    female["age"].min(),
    female["age"].max()
])

y_f = slope_f * x_f + intercept_f

plt.plot(
    x_f,
    y_f,
    color="red",
    linewidth=2,
    label="Regression Line"
)

plt.xlabel("Age")
plt.ylabel("Evaluation Score")
plt.title("Female Instructors")
plt.legend()


# ---------------- MALE ----------------
plt.subplot(1,2,2)

male = df[df["gender"] == "male"]

plt.scatter(
    male["age"],
    male["eval"],
    alpha=0.6,
    color="skyblue",
    edgecolor="black"
)

# Regression line for male
slope_m, intercept_m = np.polyfit(
    male["age"],
    male["eval"],
    1
)

x_m = np.array([
    male["age"].min(),
    male["age"].max()
])

y_m = slope_m * x_m + intercept_m

plt.plot(
    x_m,
    y_m,
    color="red",
    linewidth=2,
    label="Regression Line"
)

plt.xlabel("Age")
plt.ylabel("Evaluation Score")
plt.title("Male Instructors")
plt.legend()

plt.tight_layout()
plt.show()
```

I'm creating a side-by-side subplot to compare the relationship between age and evaluation score by gender.

For the left plot I'm filtering the dataframe for female instructors making a hotpink scatter plot and fitting a red regression line.

For the right plot I'm doing the same for male instructors with a skyblue scatter plot and its own regression line then cleaning up the layout and displaying the figure.

(Q5) Gender + Tenure differentiated scatterplot

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(10,7))

for gender in df["gender"].unique():

    for tenure in df["tenure"].unique():

        subset = df[
            (df["gender"] == gender) &
            (df["tenure"] == tenure)
        ]

        marker = "o" if tenure == "yes" else "^"

        # Scatter plot
        plt.scatter(
            subset["age"],
            subset["eval"],
            label=f"{gender}, tenure={tenure}",
            marker=marker,
            alpha=0.7
        )

        # Regression line
        if len(subset) > 1:
            slope, intercept = np.polyfit(
                subset["age"],
                subset["eval"],
                1
            )

            x = np.array([
                subset["age"].min(),
                subset["age"].max()
            ])

            y = slope * x + intercept

            plt.plot(
                x,
                y,
                linewidth=2
            )

plt.xlabel("Instructor Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Age vs Evaluation by Gender and Tenure")

plt.legend()
plt.show()
```

I'm looping through each unique combination of gender and tenure in my dataframe to create a combined scatter and regression plot.

For each group I'm assigning a specific marker (circles for tenured and triangles for non-tenured) to tell them apart plotting age versus evaluation score and then calculating and drawing a separate regression line for each category before adding the labels, title and legend to bring it all together.
